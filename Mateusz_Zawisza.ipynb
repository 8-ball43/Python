## 1. `print()` — pierwszy kontakt z Pythonem

`print()` pokazuje informację pod komórką kodu. **Tekst zapisujemy w cudzysłowie**; napis po `#` jest komentarzem.

````python
print("Witaj w Pythonie!")
print("Uczę się programowania.")
print(2 + 3)  # Komputer obliczy 2 + 3
````

## 2. Zmienne — pudełka na dane

Zmienna to **nazwa**, pod którą przechowujemy wartość, np. tekst lub liczbę.

- `imie = "Ola"` — zapisujemy tekst pod nazwą `imie`.
- `wiek = 17` — zapisujemy liczbę pod nazwą `wiek`.
- `=` oznacza **przypisanie wartości**, a nie matematyczne „równa się”.
- W nazwach zmiennych **nie używamy spacji**; Python odróżnia wielkie i małe litery.

````python
imie = "Ola"
wiek = 17
klasa = "3 TC"

print(imie)
print(wiek)
print(klasa)
````

**Zmienne można zmieniać.** Obliczenia wykonują się na aktualnej wartości.

````python
punkty = 5
print("Punkty na początku:", punkty)

punkty = punkty + 3
print("Punkty po dodaniu:", punkty)
````

## 3. Typy danych — tekst czy liczba?

| Typ | Przykład | Do czego służy? |
|---|---|---|
| `str` | `"Warszawa"` | Tekst |
| `int` | `17` | Liczba całkowita |
| `float` | `4.5` | Liczba z częścią dziesiętną (używamy **kropki**) |
| `bool` | `True` | Wartość logiczna prawda/fałsz |

`type()` pokazuje typ zmiennej.

````python
miasto = "Warszawa"
wiek = 17
srednia = 4.5
czy_uczen = True

print(type(miasto))
print(type(wiek))
print(type(srednia))
print(type(czy_uczen))
````

## 4. Obliczenia na zmiennych

Python potrafi obliczać wynik na podstawie wartości zmiennych.

`+` dodawanie · `-` odejmowanie · `*` mnożenie · `/` dzielenie · `//` dzielenie całkowite · `%` reszta z dzielenia.

````python
a = 11
b = 4

print("Suma:", a + b)
print("Różnica:", a - b)
print("Iloczyn:", a * b)
print("Dzielenie:", a / b)
print("Całkowite:", a // b)
print("Reszta:", a % b)
````

Wzory: `pole = szerokosc * wysokosc`, `obwod = 2 * (szerokosc + wysokosc)`.

````python
# ZADANIE 4 — oblicz pole i obwód, korzystając ze zmiennych.
szerokosc = 6
wysokosc = 4

print("pole:", szerokosc*wysokosc)
print("obwód",  2*szerokosc + 2*wysokosc)
````

## 5. Jak łączyć teksty?

Są trzy proste sposoby:

1. `print("Cześć", imie)` — przecinek dodaje odstęp.
2. `"Cześć " + imie` — `+` **skleja teksty**.
3. `f"Cześć {imie}"` — *f-string*, wygodniejszy sposób (więcej na drugiej godzinie).

````python
imie = "Kasia"
print("Cześć", imie)
print("Cześć " + imie + "!")
print(f"Cześć {imie}!")
````

**Uwaga na częsty błąd:** nie można połączyć tekstu i liczby przez `+` bez zamiany liczby na tekst.

❌ `"Mam " + 17 + " lat"` — błąd typów.

✅ `"Mam " + str(17) + " lat"` — działa.

````python
wiek = 17
print("Mam " + str(wiek) + " lat")
print("Mam", wiek, "lat")
````

## 6. f-string — najwygodniejsze łączenie informacji

Przed cudzysłowem wpisujemy literę **`f`**, a nazwę zmiennej umieszczamy w klamrach `{}`. Nie trzeba używać `str()`.

````python
imie = "Adam"
wiek = 17
punkty = 125

print(f"Cześć, {imie}!")
print(f"Masz {wiek} lat i {punkty} punktów.")
print(f"Za rok będziesz mieć {wiek + 1} lat.")
````

## 8. `if` i `else` — pierwsza decyzja programu

`if` sprawdza warunek. Gdy warunek nie jest spełniony, można wykonać kod po `else`.

**Uwaga:** w Pythonie ważne są wcięcia (zwykle 4 spacje).

````python
punkty = 12

if punkty >= 10:
    print("Poziom zaliczony!")
else:
    print("Spróbuj ponownie!")
````
