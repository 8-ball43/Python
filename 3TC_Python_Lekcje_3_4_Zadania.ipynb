### 2. Więcej niż dwa przypadki: `if`, `elif`, `else`

`elif` oznacza „w przeciwnym razie, jeśli…”. Warunki są sprawdzane od góry. Gdy jeden z nich jest prawdziwy, pozostałe gałęzie są pomijane.


````python
temperatura = 18

if temperatura < 0:
    print("Mróz")
elif temperatura < 15:
    print("Chłodno")
elif temperatura < 25:
    print("Ciepło")
else:
    print("Gorąco")
````

W przykładzie nie trzeba w drugim warunku pisać `temperatura >= 0`. Skoro wykonuje się `elif`, to znaczy, że wcześniejszy warunek był fałszywy.

Operatory porównania: `==`, `!=`, `>`, `<`, `>=`, `<=`.

### 6. Pętla `for` i `range()`

Pętla pozwala wykonać podobny kod wiele razy, bez kopiowania tych samych linijek.

`range(1, 6)` daje liczby `1, 2, 3, 4, 5` – wartość końcowa nie jest uwzględniana.

`range(2, 11, 2)` daje `2, 4, 6, 8, 10`; trzeci argument oznacza krok.


````python
for numer in range(1, 6):
    print("Próba numer", numer)

print("Liczby parzyste:")
for liczba in range(2, 11, 2):
    print(liczba)
````

### 7. Licznik i suma w pętli

Zmienną można aktualizować w każdej iteracji. Zapis `suma += liczba` to skrót od `suma = suma + liczba`.

### 8. Pętla `while` – powtarzaj, dopóki warunek jest spełniony

`for` jest wygodne, kiedy wiemy, ile razy chcemy coś powtórzyć. `while` przydaje się wtedy, gdy liczba powtórzeń nie jest z góry znana.

**Uwaga:** warunek musi kiedyś stać się fałszywy albo program musi dojść do `break`. W przeciwnym razie pętla będzie działać bez końca. W Jupyter można ją zatrzymać przyciskiem przerwania wykonywania komórki.


````python
licznik = 3
while licznik > 0:
    print("Odliczanie:", licznik)
    licznik -= 1
print("Start!")
````

`break` natychmiast przerywa pętlę. Można go użyć np. gdy użytkownik wpisze polecenie zakończenia.


````python
while True:
    tekst = input("Wpisz komunikat lub 'koniec': ")
    if tekst == "koniec":
        break
    print("Wpisałeś:", tekst)
print("Program zakończony")
````

### Zadanie 10. Trzy próby podania hasła (5 min)

Użytkownik może maksymalnie trzy razy wpisać hasło `informatyka`.

Jeśli wpisze prawidłowe hasło, program wyświetla `Dostęp przyznany` i kończy próby. Jeśli po trzech próbach hasło nadal jest błędne, wyświetla `Dostęp zablokowany`.

Nie używaj trzech oddzielnych `input()` – wykorzystaj pętlę.


````python
# Zadanie 10
poprawne_haslo = "informatyka"
# Wykorzystaj for lub while i break.
licznik = 0
while licznik < 3:
    licznik = licznik + 1
    print(licznik)
    podane = input("Podaj hasło:")
    if podane == poprawne_haslo:
        print("Dostep przyznany")
        break
    if licznik == 3:
        print("Dostęp zablokowany")
````

### 9. Listy – wiele wartości w jednej zmiennej

Lista przechowuje dane w określonej kolejności. Elementy zapisujemy w nawiasach kwadratowych. Pierwszy element ma indeks `0`.

Najczęściej używane operacje:
- `append()` – dodaje nowy element;
- `len()` – podaje liczbę elementów;
- `sum()` – sumuje liczby;
- `lista[0]` – odczytuje pierwszy element;
- `pop(indeks)` – usuwa element o podanym indeksie (użyjemy w zadaniu dodatkowym).


````python
oceny = [3, 4, 5]
oceny.append(4)

print("Pierwsza ocena:", oceny[0])
print("Wszystkie oceny:")
for ocena in oceny:
    print(ocena)

print("Ile ocen:", len(oceny))
print("Średnia:", sum(oceny) / len(oceny))
````
