## 1. Słownik (`dict`)

Lista pozwala przechowywać kilka wartości pod kolejnymi indeksami. Gdy jedna rzecz ma wiele cech, wygodniej użyć **słownika**: każda wartość ma własny klucz.

Przykładowy produkt ma nazwę, cenę i liczbę sztuk.


```python
produkt = {"nazwa": "Klawiatura", "cena": 119.90, "sztuki": 4}

print(produkt["nazwa"])
print(f"Wartość zapasu: {produkt['cena'] * produkt['sztuki']:.2f} zł")

produkt["sztuki"] -= 1
produkt["kategoria"] = "Akcesoria"
print(produkt)
```

Ważne:
- `produkt["nazwa"]` – odczyt wartości pod kluczem `nazwa`.
- `produkt["sztuki"] = 5` – zmiana istniejącej wartości.
- `produkt["stan"] = "dostępny"` – dodanie nowej pary klucz–wartość.
- `produkt.get("opis", "Brak opisu")` – bezpieczny odczyt; brak klucza nie powoduje błędu.

## 2. Przechodzenie po słowniku

Metoda `.items()` daje dostęp do klucza i odpowiadającej mu wartości. Możemy użyć jej w pętli.


```python
pracownik = {"imie": "Adam", "stanowisko": "Tester", "godziny": 32}

for klucz, wartosc in pracownik.items():
    print(f"{klucz}: {wartosc}")

print(pracownik.get("telefon", "Nie podano telefonu"))
```

### Zadanie 3. Znajdź urządzenie (6 min)

Napisz funkcję `znajdz_urzadzenie(urzadzenia, nazwa)`.

Funkcja przeszukuje listę słowników i zwraca cały słownik odpowiadający podanej nazwie. Wielkość liter nie ma znaczenia. Jeżeli urządzenia nie znaleziono, zwróć `None`.

Sprawdź wyszukiwanie `"laptop"` i `"drukarka"` na liście z przykładu. Nie używaj `input()` w funkcji.


```python
# Zadanie 3
urzadzenia = [
    {"nazwa": "Laptop", "wolne": 3},
    {"nazwa": "Projektor", "wolne": 0},
    {"nazwa": "Tablet", "wolne": 5}
]

def znajdz_urzadzenie(urzadzenia, nazwa):
    # Porównuj nazwy po zamianie liter na małe.
    for urzadzenie in urzadzenia:
        if urzadzenie['nazwa'] == nazwa:
            return urzadzenie['nazwa']
    return None


print(znajdz_urzadzenie(urzadzenia, "Laptop"))

# Przetestuj funkcję na kilku nazwach.
```

### Zadanie 4. Raport magazynu (5 min)

Napisz funkcję `raport_magazynu(urzadzenia)`, która zwraca napis w postaci:

`Pozycji: 3, sztuk dostępnych: 8, braków: 1`

Pole `braków` oznacza liczbę **rodzajów urządzeń**, których stan wynosi 0, a nie brakujące sztuki. Dla pustej listy wynik to: `Pozycji: 0, sztuk dostępnych: 0, braków: 0`.


```python
# Zadanie 4
urzadzenia = [
    {"nazwa": "Laptop", "wolne": 3},
    {"nazwa": "Projektor", "wolne": 0},
    {"nazwa": "Tablet", "wolne": 5}
]

def raport_magazynu(urzadzenia):
    index = 0
    braki = 0
    for urzadzenie in urzadzenia:
        if urzadzenie['wolne'] == 0:
            braki += 1 
        index += 1
        return f"Pozycji:{index}, sztuk dostępnych: {urzadzenie['wolne']} braków: {braki}"
    return f"Pozycji: 0, sztuk dostępnych: 0 braków: 0"

print(raport_magazynu(urzadzenia))
    

# Uruchom funkcję dla listy z przykładu i dla pustej listy.
```

## 4. Zapis danych – format JSON

Zwykła lista albo słownik istnieją tylko podczas pracy programu. Jeżeli program ma pamiętać dane po ponownym uruchomieniu, trzeba je zapisać do pliku.

JSON to format tekstowy. Możemy zapisać w nim listę słowników i później ją odczytać. Użyjemy modułu `json`, który jest już w Pythonie.

Zapis `with open(...)` otwiera plik i automatycznie zamyka go po wykonaniu kodu.


```python
import json

ustawienia = {"motyw": "jasny", "powiadomienia": True}

tekst_json = json.dumps(ustawienia, ensure_ascii=False, indent=2)
print(tekst_json)
```

```python
import json

lista_do_zapisu = [
    {"nazwa": "Laptop", "wolne": 3},
    {"nazwa": "Projektor", "wolne": 0}
]

with open("przyklad_urzadzenia.json", "w", encoding="utf-8") as plik:
    json.dump(lista_do_zapisu, plik, ensure_ascii=False, indent=2)

with open("przyklad_urzadzenia.json", "r", encoding="utf-8") as plik:
    odczytane = json.load(plik)

print("Odczytane dane:", odczytane)
```

Po uruchomieniu kodu powstanie plik `przyklad_urzadzenia.json` w bieżącym katalogu roboczym Pythona. Otwórz go w edytorze i sprawdź jego zawartość. Pamiętaj, że otwarcie pliku w trybie `"w"` **nadpisuje** jego wcześniejszą zawartość.

Plik JSON jest zwykłym tekstem, ale obowiązuje w nim ustalona składnia. Nie musisz ręcznie dopisywać nawiasów czy przecinków – robi to `json.dump()`.

## 5. Co zrobić, kiedy pliku nie ma?

Program nie powinien kończyć się błędem, jeśli plik jeszcze nie powstał. `try/except` pozwala obsłużyć spodziewany problem i zdecydować, co zrobić dalej.


```python
import json

def wczytaj_liste(nazwa_pliku):
    try:
        with open(nazwa_pliku, "r", encoding="utf-8") as plik:
            return json.load(plik)
    except FileNotFoundError:
        print("Brak zapisanego pliku. Zaczynamy od pustej listy.")
        return []
    except json.JSONDecodeError:
        print("Plik ma nieprawidłową zawartość JSON.")
        return []

print(wczytaj_liste("plik_ktorego_nie_ma_12345.json"))
```
