## 2. Własna funkcja

Jeżeli kilka razy potrzebujemy wykonać podobne obliczenie, nie trzeba kopiować tego samego kodu. Możemy nadać mu nazwę i wywoływać w różnych miejscach programu.

Funkcję tworzymy za pomocą `def`. W nawiasach podajemy parametry, czyli dane potrzebne funkcji. `return` oznacza wynik, który funkcja oddaje do dalszego wykorzystania.


```python
def policz_brutto(cena_netto, vat):
    return round(cena_netto * (1 + vat), 2)

print(policz_brutto(100, 0.23))
print(policz_brutto(80, 0.08))

cena_monitora = policz_brutto(650, 0.23)
print(f"Cena monitora: {cena_monitora:.2f} zł")
```

Ważna różnica:
- `print()` tylko wyświetla napis lub liczbę.
- `return` zwraca wynik. Można go zapisać w zmiennej, porównać z inną wartością albo przekazać do kolejnej funkcji.

Jedna funkcja powinna mieć w miarę możliwości jedno konkretne zadanie.


### Zadanie 2. Pole trójkąta (4 min)

Napisz funkcję `pole_trojkata(podstawa, wysokosc)`, która **zwraca** pole trójkąta. Jeśli któryś wymiar jest mniejszy lub równy zero, funkcja powinna zwrócić `None`.

Sprawdź działanie dla danych `(8, 5)`, `(3, 10)` i `(-2, 5)`. Prawidłowe wyniki: `20.0`, `15.0`, `None`.


```python
# Zadanie 2
def pole_trojkata(podstawa, wysokosc):
    # Uzupełnij warunek i obliczenie.
    pole = float((podstawa * wysokosc) / 2)
    if podstawa <= 0 or wysokosc <= 0:
        return None

    else:
     return pole
    pass

# Wywołaj funkcję dla trzech zestawów danych.
zestaw_a = pole_trojkata(8, 5)
zestaw_b = pole_trojkata(3, 10)
zestaw_c = pole_trojkata(-2, 5)
print("Pole:", zestaw_a)
print("Pole:", zestaw_b)
print("Pole:", zestaw_c)
```

### Zadanie 3. Wynik sprawdzianu (5 min)

Napisz funkcję `ocen_wynik(punkty)` zwracającą tekst:

| Punkty | Wynik |
|---|---|
| poniżej 0 lub powyżej 100 | `Błędny wynik` |
| 0–49 | `Niezaliczony` |
| 50–74 | `Zaliczony` |
| 75–89 | `Dobry wynik` |

Przetestuj wartości `-1`, `49`, `50`, `75`, `90` i `101`. Funkcja **nie pobiera danych przez `input()`** – otrzymuje je przez parametr.


```python
# Zadanie 3
def ocen_wynik(punkty):
    if punkty < 49:
        return "Niezaliczony"
    elif punkty < 75:
        return "Zaliczony"
    elif punkty < 90:
        return "Dobry Wynik"
    elif punkty >= 90:
        return "Bardzo dobry wynik"
    elif punkty < 0 or punkty > 100:
        return "Błędny wynik"
sprawdzenie = ocen_wynik(101)
print(sprawdzenie)

# Testy: wyświetl wynik działania funkcji dla podanych liczb.
```

### Zadanie 4. Koszt wydruku (4 min)

Napisz funkcję `koszt_wydruku(strony, cena_strony=0.40)`.

- Zwraca cenę wydruku zaokrągloną do dwóch miejsc.
- Jeżeli liczba stron jest ujemna albo cena strony jest ujemna, zwraca `None`.
- Sprawdź: 15 stron w cenie domyślnej, 15 stron po 0.70 zł i -2 strony.

Nie używaj `input()` wewnątrz funkcji.


```python
# Zadanie 4
def koszt_wydruku(strony, cena_strony=0.40):
    cena = strony * cena_strony
    cena = round(cena, 2)
    if cena < 0 :
        return None
    else:
        return cena
zamowienie = koszt_wydruku(-2)
print(zamowienie)
# Sprawdź trzy przypadki.
```

## 4. Funkcja może pracować na liście

Listę też można przekazać jako argument. Funkcja może przejść po jej elementach i zwrócić nową listę albo jedną liczbę.


```python
def wyniki_zaliczone(wyniki, prog=50):
    zaliczone = []
    for wynik in wyniki:
        if wynik >= prog:
            zaliczone.append(wynik)
    return zaliczone

punkty = [20, 64, 89, 47, 76]
print(wyniki_zaliczone(punkty))
print(wyniki_zaliczone(punkty, 75))
print("Lista początkowa:", punkty)
```

### Zadanie 5. Wyszukiwanie w liście (5 min)

Napisz funkcję `znajdz_nazwy(nazwy, fragment)`, która zwraca nową listę zawierającą nazwy z podanym fragmentem. Wielkość liter nie ma znaczenia.

Dla `nazwy = ["Python", "Java", "JavaScript", "C#", "Django"]` i fragmentu `"ja"` funkcja powinna zwrócić `["Java", "JavaScript"]`.

Wskazówka: `"py" in "python"` daje `True`.


```python
# Zadanie 5
def znajdz_nazwy(nazwy, fragment):
    wyniki = []
    for nazwa in nazwy:
        if fragment.lower() in nazwa.lower():
         wyniki.append(nazwa)
        
    # Przejdź po nazwach i sprawdź, czy zawierają fragment.
    return wyniki

nazwy = ["Python", "Java", "JavaScript", "C#", "Django"]
wywolanie_a = znajdz_nazwy(nazwy, "ja")
wywolanie = znajdz_nazwy(nazwy, "py")
print(wywolanie)
print(wywolanie_a)
# Wywołaj funkcję dla fragmentów "ja" i "PY".
```

### Zadanie 6. Bez powtarzających się wartości (5 min)

Napisz funkcję `usun_duplikaty(liczby)`. Zwróć nową listę bez powtórzeń, zachowując kolejność pierwszego wystąpienia każdej liczby. Nie zmieniaj oryginalnej listy.

Przykład: `[2, 5, 2, 8, 5, 3]` → `[2, 5, 8, 3]`.

Wskazówka: zanim użyjesz `append()`, sprawdź, czy danej liczby nie ma już w nowej liście.


```python
# Zadanie 6
def usun_duplikaty(liczby):
    wynik = []
    # Dokończ.
    for liczba in liczby:
        if liczba not in liczby:
            wyniki.append(liczba)
    return wynik
liczby = [2, 5, 2, 8, 5, 3]
print(usun_duplikaty(liczby))
# Sprawdź funkcję na liście [2, 5, 2, 8, 5, 3].
```

## 5. Sprawdzanie funkcji

Funkcje wygodnie testować na kilku danych bez ciągłego wpisywania ich z klawiatury. Polecenie `assert` sprawdza, czy podany warunek jest prawdziwy. Jeżeli nie jest, program zgłosi błąd.


```python
def czy_dodatnia(liczba):
    return liczba > 0

assert czy_dodatnia(10) == True
assert czy_dodatnia(0) == False
assert czy_dodatnia(-5) == False

print("Wszystkie testy przeszły poprawnie.")
```

### Zadanie 7. Analiza wyników (5 min)

Napisz funkcję `raport_wynikow(wyniki)`. Dla pustej listy zwróć `"Brak wyników"`, a dla pozostałych wyników zwróć napis z liczbą wyników, najwyższym wynikiem oraz średnią zaokrągloną do jednego miejsca.

Przykład dla `[50, 75, 100]`:

`Liczba: 3, maksimum: 100, średnia: 75.0`

Po napisaniu funkcji dodaj trzy własne testy `assert`, również dla pustej listy.


```python
# Zadanie 7
def raport_wynikow(wyniki):
    if len(wyniki) == 0:
     return "Brak wyników"
    else :
     return "Licbza: ", len(wyniki), max(wyniki), min(wyniki), "średnia: ", round((sum(wyniki) / len(wyniki)),1)
oceny = [50,75,100]
print(raport_wynikow(oceny))

assert raport_wynikow(oceny) == "Liczba: 3, 100, 50, średnia: 75.0"
# Sprawdź listę [50, 75, 100], listę [10] i pustą listę.
# Następnie zapisz testy assert.
```
